```python
customer_orders = pd.merge(left=Customers,right=Orders,on='customer_id' , how='left' )
```

```python
customer_orders.info()
```

```python
# converting Dtype of order_purchase_timestamp to datetime 
customer_orders['order_purchase_timestamp'] = pd.to_datetime(customer_orders['order_purchase_timestamp'])

# Find first order date per customer
first_orders = customer_orders.groupby('customer_id')['order_purchase_timestamp'].min().reset_index()

# Extract year-month for first purchase
first_orders['month'] = first_orders['order_purchase_timestamp'].dt.to_period('M')

# Count new customers per month
new_customers_by_month = first_orders['month'].value_counts().sort_index()

# Convert to DataFrame for display
new_customers_df = new_customers_by_month.reset_index()
new_customers_df.columns = ['Month', 'New_Customers']

print(new_customers_df)
```

```python
# Convert to datetime
Orders['order_purchase_timestamp'] = pd.to_datetime(Orders['order_purchase_timestamp'])

# Create 'OrderMonth' and 'group_belongs'
Orders['OrderMonth'] = Orders['order_purchase_timestamp'].dt.to_period('M')
Orders['group_belongs'] = Orders.groupby('customer_id')['order_purchase_timestamp'].transform('min').dt.to_period('M')
```

```python
from operator import attrgetter 
# Calculate group index (how many months since first purchase)
Orders['group_index'] = (Orders['OrderMonth'] - Orders['group_belongs']).apply(attrgetter('n'))
```

```python
# Count customers per group and group index
group_data = Orders.groupby(['group_belongs', 'group_index'])['customer_id'].nunique().reset_index()

# Pivot table: Retention matrix
group_pivot = group_data.pivot(index='group_belongs', columns='group_index', values='customer_id')
```

```python
group_size = group_pivot.iloc[:,0]
retention = group_pivot.divide(group_size, axis=0).round(3)

print("Retention Table (Month-over-Month %):")
print(retention.fillna(0) * 100)
```

## Q-3

```python
from itertools import combinations
from collections import Counter

# Group products by order_id
order_products = Order_item.groupby('order_id')['product_id'].apply(list)

# Filter orders with 2 or more products
order_products = order_products[order_products.apply(lambda x: len(x) >= 2)]

# Generate all combinations of size 2 or 3
all_combinations = []
for products in order_products:
    combos = list(combinations(sorted(products), 2))  # use 3 for triplets if needed
    all_combinations.extend(combos)

# Count frequency of each combination
combo_counts = Counter(all_combinations)

# Convert to DataFrame and get top 10
top_combos = pd.DataFrame(combo_counts.most_common(10), columns=['Product_Combination', 'Frequency'])

print(top_combos)
```
